Now that the data is cleaned and ready we can merge all the datasets to create a master dataset which can be used for further analysis and data modeling

```python
#Using duckDB to merge all the datasets into a single master dataset
import duckdb
conn = duckdb.connect('master_dataset.db')
conn.register('df_cp', df_cp)
conn.register('df_di', df_di)
conn.register('df_t', df_t)

master = conn.execute("""
WITH t_agg AS (
  SELECT customer_id,
         COUNT(*)           AS txn_count,
         SUM(Amount_INR)    AS txn_sum,
         AVG(Amount_INR)    AS txn_avg,
         MIN(Amount_INR)    AS txn_min,
         MAX(Amount_INR)    AS txn_max
  FROM df_t
  GROUP BY customer_id
)
SELECT *
FROM df_cp
INNER JOIN df_di USING (customer_id)
INNER JOIN t_agg   USING (customer_id)
""").fetch_df()

conn.close()
master.to_csv('master_dataset.csv', index=False)
```
